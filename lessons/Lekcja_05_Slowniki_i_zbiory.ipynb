# Lekcja 5: Słowniki i zbiory

W tej lekcji poznasz dwie zaawansowane struktury danych: słowniki (dict) i zbiory (set).

## 1. Słowniki - podstawy

Słownik to kolekcja par klucz-wartość. Tworzymy go używając nawiasów klamrowych `{}`.

```python
# Tworzenie słowników
pusty_slownik = {}

osoba = {
    "imie": "Jan",
    "nazwisko": "Kowalski",
    "wiek": 30,
    "miasto": "Warszawa"
}

print("Pusty słownik:", pusty_slownik)
print("Osoba:", osoba)
print(f"Typ: {type(osoba)}")

# Inny sposób tworzenia
slownik2 = dict(imie="Anna", wiek=25)
print("\nSłownik2:", slownik2)
```

## 2. Dostęp do wartości

```python
osoba = {
    "imie": "Jan",
    "nazwisko": "Kowalski",
    "wiek": 30,
    "miasto": "Warszawa"
}

# Dostęp przez klucz
print("Imię:", osoba["imie"])
print("Wiek:", osoba["wiek"])

# Metoda get() - bezpieczniejsza
print("\nMiasto:", osoba.get("miasto"))
print("Zawód:", osoba.get("zawod"))  # Zwraca None jeśli klucz nie istnieje
print("Zawód:", osoba.get("zawod", "Nie podano"))  # Domyślna wartość


# Różnica między [] a get()
# print(osoba["zawod"])  # To spowoduje błąd KeyError!
```

## 3. Modyfikowanie słowników

```python
osoba = {"imie": "Jan", "wiek": 30}
print("Początkowy słownik:", osoba)

# Dodawanie nowego klucza
osoba["miasto"] = "Kraków"
print("Po dodaniu miasta:", osoba)

# Modyfikacja wartości
osoba["wiek"] = 31
print("Po zmianie wieku:", osoba)

# Dodawanie wielu elementów - update()
osoba.update({"zawod": "programista", "email": "jan@example.com"})
print("Po update():", osoba)

# Usuwanie elementów
del osoba["email"]
print("\nPo del:", osoba)

# pop() - usuwa i zwraca wartość
zawod = osoba.pop("zawod")
print(f"Po pop('zawod'): {osoba}")
print(f"Usunięty zawód: {zawod}")

# popitem() - usuwa i zwraca ostatnią parę
ostatni = osoba.popitem()
print(f"\nPo popitem(): {osoba}")
print(f"Usunięta para: {ostatni}")

# clear() - usuwa wszystkie elementy
temp = {"a": 1, "b": 2}
temp.clear()
print(f"Po clear(): {temp}")
```

## 4. Metody słowników

```python
osoba = {
    "imie": "Jan",
    "nazwisko": "Kowalski",
    "wiek": 30,
    "miasto": "Warszawa"
}

# keys() - zwraca klucze
print("Klucze:", osoba.keys())
print("Lista kluczy:", list(osoba.keys()))

# values() - zwraca wartości
print("\nWartości:", osoba.values())
print("Lista wartości:", list(osoba.values()))

# items() - zwraca pary (klucz, wartość)
print("\nPary klucz-wartość:")
for klucz, wartosc in osoba.items():
    print(f"  {klucz}: {wartosc}")

# Sprawdzanie czy klucz istnieje
print("\n'imie' in osoba:", "imie" in osoba)
print("'zawod' in osoba:", "zawod" in osoba)

# Długość słownika
print(f"\nLiczba elementów: {len(osoba)}")
```

## 5. Iterowanie po słownikach

```python
oceny = {
    "matematyka": 5,
    "fizyka": 4,
    "chemia": 5,
    "informatyka": 6
}

# Iterowanie po kluczach (domyślnie)
print("Przedmioty:")
for przedmiot in oceny.keys():
    print(f"  {przedmiot}")

# Iterowanie po wartościach
print("\nOceny:")
for ocena in oceny.values():
    print(f"  {ocena}")

# Iterowanie po parach klucz-wartość
print("\nPrzedmiot - Ocena:")
for przedmiot, ocena in oceny.items():
    print(f"  {przedmiot}: {ocena}")

# Obliczanie średniej
srednia = sum(oceny.values()) / len(oceny)
print(f"\nŚrednia ocen: {srednia:.2f}")
```

## 6. Zagnieżdżone słowniki

```python
# Słownik słowników
studenci = {
    "student1": {
        "imie": "Jan",
        "wiek": 20,
        "oceny": [5, 4, 5]
    },
    "student2": {
        "imie": "Anna",
        "wiek": 21,
        "oceny": [6, 5, 5]
    },
    "student3": {
        "imie": "Piotr",
        "wiek": 19,
        "oceny": [4, 4, 3]
    }
}

# Dostęp do zagnieżdżonych wartości
print("Imię student1:", studenci["student1"]["imie"])
print("Wiek student2:", studenci["student2"]["wiek"])
print("Oceny student3:", studenci["student3"]["oceny"])

# Iterowanie po zagnieżdżonym słowniku
print("\nWszyscy studenci:")
for id_studenta, dane in studenci.items():
    print(f"\n{id_studenta}:")
    print(f"  Imię: {dane['imie']}")
    print(f"  Wiek: {dane['wiek']}")
    srednia = sum(dane['oceny']) / len(dane['oceny'])
    print(f"  Średnia ocen: {srednia:.2f}")
```

## 7. Dictionary comprehensions

```python
# Podstawowe dictionary comprehension
kwadraty = {x: x**2 for x in range(1, 6)}
print("Kwadraty:", kwadraty)

# Z warunkiem
parzyste_kwadraty = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print("Parzyste kwadraty:", parzyste_kwadraty)

# Zamiana kluczy i wartości
oryginalny = {"a": 1, "b": 2, "c": 3}
odwrocony = {v: k for k, v in oryginalny.items()}
print("\nOryginalny:", oryginalny)
print("Odwrócony:", odwrocony)

# Przekształcanie stringów
slowa = ["python", "java", "c++"]
dlugosci = {slowo: len(slowo) for slowo in slowa}
print("\nDługości słów:", dlugosci)

# Z if-else
liczby = [1, 2, 3, 4, 5]
etykiety = {x: "parzysta" if x % 2 == 0 else "nieparzysta" for x in liczby}
print("\nEtykiety:", etykiety)
```

## 8. Zbiory (Sets) - podstawy

Zbiór to nieuporządkowana kolekcja unikalnych elementów.

```python
# Tworzenie zbiorów
pusty_zbior = set()  # Uwaga: {} tworzy pusty słownik, nie zbiór!
liczby = {1, 2, 3, 4, 5}
owoce = {"jabłko", "banan", "pomarańcza"}

print("Pusty zbiór:", pusty_zbior)
print("Liczby:", liczby)
print("Owoce:", owoce)

# Zbiór automatycznie usuwa duplikaty
liczby_z_duplikatami = {1, 2, 2, 3, 3, 3, 4, 5}
print("\nZ duplikatami:", liczby_z_duplikatami)

# Tworzenie zbioru z listy
lista = [1, 2, 2, 3, 3, 3, 4, 5]
zbior = set(lista)
print("Ze listy:", zbior)

# Długość zbioru
print(f"\nDługość: {len(liczby)}")
```

## 9. Operacje na zbiorach

```python
zbior = {1, 2, 3}
print("Początkowy zbiór:", zbior)

# Dodawanie elementu
zbior.add(4)
print("Po add(4):", zbior)

# Dodawanie wielu elementów
zbior.update([5, 6, 7])
print("Po update([5, 6, 7]):", zbior)

# Usuwanie elementu - remove() (błąd jeśli nie istnieje)
zbior.remove(7)
print("Po remove(7):", zbior)

# Usuwanie elementu - discard() (brak błędu jeśli nie istnieje)
zbior.discard(10)  # Nie ma 10, ale nie ma błędu
print("Po discard(10):", zbior)

# pop() - usuwa i zwraca losowy element
element = zbior.pop()
print(f"\nPo pop(): {zbior}")
print(f"Usunięty element: {element}")

# clear() - usuwa wszystkie elementy
temp = {1, 2, 3}
temp.clear()
print(f"Po clear(): {temp}")

# Sprawdzanie przynależności
liczby = {1, 2, 3, 4, 5}
print(f"\n3 in liczby: {3 in liczby}")
print(f"10 in liczby: {10 in liczby}")
```

## 10. Operacje matematyczne na zbiorach

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Zbiór A:", A)
print("Zbiór B:", B)

# Suma (union) - wszystkie elementy z obu zbiorów
suma = A | B
print("\nA | B (suma):", suma)
print("A.union(B):", A.union(B))

# Przecięcie (intersection) - wspólne elementy
przeciecie = A & B
print("\nA & B (przecięcie):", przeciecie)
print("A.intersection(B):", A.intersection(B))

# Różnica (difference) - elementy w A, ale nie w B
roznica = A - B
print("\nA - B (różnica):", roznica)
print("A.difference(B):", A.difference(B))

# Różnica symetryczna - elementy w A lub B, ale nie w obu
roznica_sym = A ^ B
print("\nA ^ B (różnica symetryczna):", roznica_sym)
print("A.symmetric_difference(B):", A.symmetric_difference(B))
```

## 11. Relacje między zbiorami

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}
C = {6, 7, 8}

# Podzbiór (subset)
print(f"A jest podzbiorem B: {A.issubset(B)}")
print(f"A <= B: {A <= B}")

# Nadzbiór (superset)
print(f"\nB jest nadzbiorem A: {B.issuperset(A)}")
print(f"B >= A: {B >= A}")

# Rozłączne (disjoint) - brak wspólnych elementów
print(f"\nA i C są rozłączne: {A.isdisjoint(C)}")
print(f"A i B są rozłączne: {A.isdisjoint(B)}")
```

## 12. Frozenset - niezmienny zbiór

```python
# Frozenset to niezmienny zbiór
zbior_zmienny = {1, 2, 3}
zbior_niezmienny = frozenset([1, 2, 3])

print("Zmienny:", zbior_zmienny)
print("Niezmienny:", zbior_niezmienny)

# Można go używać jako klucza w słowniku
slownik = {
    frozenset([1, 2]): "para 1,2",
    frozenset([3, 4]): "para 3,4"
}
print("\nSłownik z frozenset:", slownik)

# Nie można modyfikować
# zbior_niezmienny.add(4)  # To spowoduje błąd!
```

## 13. Praktyczne zastosowania

```python
# 1. Usuwanie duplikatów z listy
lista_z_duplikatami = [1, 2, 2, 3, 3, 3, 4, 5, 5]
bez_duplikatow = list(set(lista_z_duplikatami))
print("Bez duplikatów:", bez_duplikatow)

# 2. Sprawdzanie unikalnych liter w słowie
slowo = "programowanie"
unikalne_litery = set(slowo)
print(f"\nUnikalne litery w '{slowo}': {unikalne_litery}")
print(f"Liczba unikalnych liter: {len(unikalne_litery)}")

# 3. Znajdowanie wspólnych elementów
lista1 = [1, 2, 3, 4, 5]
lista2 = [4, 5, 6, 7, 8]
wspolne = set(lista1) & set(lista2)
print(f"\nWspólne elementy: {wspolne}")

# 4. Licznik słów (prosty)
tekst = "python jest super python jest fajny"
slowa = tekst.split()
licznik = {}
for slowo in slowa:
    licznik[slowo] = licznik.get(slowo, 0) + 1
print(f"\nLicznik słów: {licznik}")
```
