## 1. Czym jest klasa i obiekt?

- **Klasa** to szablon/przepis na tworzenie obiektów
- **Obiekt** to konkretna instancja klasy

Analogia: Klasa to przepis na ciasto, obiekt to konkretne upieczone ciasto.

```python
# Najprostsza klasa
class Pies:
    pass  # Pusta klasa

# Tworzenie obiektów (instancji)
pies1 = Pies()
pies2 = Pies()

print(f"pies1: {pies1}")
print(f"pies2: {pies2}")
print(f"Czy to ten sam obiekt? {pies1 is pies2}")
```

## 2. Atrybuty (właściwości)

```python
class Pies:
    # Atrybut klasowy (wspólny dla wszystkich instancji)
    gatunek = "Canis familiaris"

# Tworzenie obiektów i dodawanie atrybutów instancji
pies1 = Pies()
pies1.imie = "Burek"
pies1.wiek = 5

pies2 = Pies()
pies2.imie = "Azor"
pies2.wiek = 3
pies2.ksywa = "Bobik"

print(f"{pies1.imie} ma {pies1.wiek} lat, gatunek: {pies1.gatunek}")
print(f"{pies2.imie} ma {pies2.wiek} lata, gatunek: {pies2.gatunek}")
print(f"{pies2.imie} ma {pies2.wiek} lata, gatunek: {pies2.gatunek} ksywa: {pies2.ksywa}.")
```

## 3. Konstruktor (__init__)

Konstruktor to specjalna metoda wywoływana podczas tworzenia obiektu.

```python
class Pies:
    def __init__(self, imie, wiek):
        """Konstruktor klasy Pies."""
        self.imie = imie  # Atrybut instancji
        self.wiek = wiek
        print(f"Utworzono psa: {imie}")

# Teraz musimy podać argumenty przy tworzeniu
pies1 = Pies("Burek", 5)
pies2 = Pies("Azor", 3)

print(f"\n{pies1.imie} ma {pies1.wiek} lat")
print(f"{pies2.imie} ma {pies2.wiek} lata")
```

## 4. Metody (funkcje w klasie)

```python
class Pies:
    def __init__(self, imie, wiek):
        self.imie = imie
        self.wiek = wiek
    
    def szczekaj(self):
        """Pies szczeka."""
        return f"{self.imie} mówi: Hau hau!"
    
    def opis(self):
        """Zwraca opis psa."""
        return f"{self.imie} ma {self.wiek} lat"
    
    def urodziny(self):
        """Zwiększa wiek psa o 1."""
        self.wiek += 1
        return f"{self.imie} ma teraz {self.wiek} lat!"

# Używanie metod
pies = Pies("Burek", 5)
print(pies.szczekaj())
print(pies.opis())
print(pies.urodziny())
print(pies.opis())
```

## 5. Metoda __str__ (reprezentacja tekstowa)

```python
class Pies:
    def __init__(self, imie, wiek):
        self.imie = imie
        self.wiek = wiek
    
    def __str__(self):
        """Zwraca czytelną reprezentację obiektu."""
        return f"Pies(imię={self.imie}, wiek={self.wiek})"
    
    def szczekaj(self):
        return f"{self.imie}: Hau hau!"

pies = Pies("Burek", 5)
print(pies)  # Automatycznie wywołuje __str__
print(pies.szczekaj())
```

## 6. Przykład: Klasa Konto Bankowe

```python
class KontoBankowe:
    def __init__(self, wlasciciel, saldo=0):
        self.wlasciciel = wlasciciel
        self.saldo = saldo
        self.historia = []
    
    def wplata(self, kwota):
        """Wpłaca pieniądze na konto."""
        if kwota > 0:
            self.saldo += kwota
            self.historia.append(f"Wpłata: +{kwota} zł")
            return f"Wpłacono {kwota} zł. Nowe saldo: {self.saldo} zł"
        return "Kwota musi być dodatnia!"
    
    def wyplata(self, kwota):
        """Wypłaca pieniądze z konta."""
        if kwota > self.saldo:
            return "Niewystarczające środki!"
        if kwota > 0:
            self.saldo -= kwota
            self.historia.append(f"Wypłata: -{kwota} zł")
            return f"Wypłacono {kwota} zł. Nowe saldo: {self.saldo} zł"
        return "Kwota musi być dodatnia!"
    
    def pokaz_saldo(self):
        """Pokazuje aktualne saldo."""
        return f"Saldo konta {self.wlasciciel}: {self.saldo} zł"
    
    def pokaz_historie(self):
        """Pokazuje historię transakcji."""
        print(f"Historia konta {self.wlasciciel}:")
        for operacja in self.historia:
            print(f"  {operacja}")
    
    def __str__(self):
        return f"Konto({self.wlasciciel}, saldo={self.saldo} zł)"

# Użycie klasy
konto = KontoBankowe("Jan Kowalski", 1000)
print(konto)
print()

print(konto.wplata(500))
print(konto.wyplata(200))
print(konto.wyplata(2000))  # Za dużo
print()

print(konto.pokaz_saldo())
print()
konto.pokaz_historie()
```

## 7. Dziedziczenie

Dziedziczenie pozwala tworzyć nowe klasy na podstawie istniejących.

```python
# Klasa bazowa (rodzic)
class Zwierze:
    def __init__(self, imie, wiek):
        self.imie = imie
        self.wiek = wiek
    
    def opis(self):
        return f"{self.imie} ma {self.wiek} lat"
    
    def dzwiek(self):
        return "Jakiś dźwięk"

# Klasy pochodne (dzieci)
class Pies(Zwierze):
    def dzwiek(self):
        return "Hau hau!"
    
    def aportuj(self):
        return f"{self.imie} aportuje piłkę!"

class Kot(Zwierze):
    def dzwiek(self):
        return "Miau!"
    
    def mruczenie(self):
        return f"{self.imie} mruczy: mrrr..."

# Użycie
pies = Pies("Burek", 5)
kot = Kot("Mruczek", 3)

print(pies.opis())
print(pies.dzwiek())
print(pies.aportuj())
print()

print(kot.opis())
print(kot.dzwiek())
print(kot.mruczenie())
```

## 8. Polimorfizm

```python
# Różne obiekty mogą mieć tę samą metodę
zwierzeta = [
    Pies("Burek", 5),
    Kot("Mruczek", 3),
    Pies("Azor", 7),
    Kot("Puszek", 2)
]

print("Wszystkie zwierzęta wydają dźwięki:")
for zwierze in zwierzeta:
    print(f"{zwierze.imie}: {zwierze.dzwiek()}")
```

## 9. Rozszerzanie konstruktora w klasie pochodnej

```python
class Zwierze:
    def __init__(self, imie, wiek):
        self.imie = imie
        self.wiek = wiek

class Pies(Zwierze):
    def __init__(self, imie, wiek, rasa):
        super().__init__(imie, wiek)  # Wywołanie konstruktora klasy bazowej
        self.rasa = rasa  # Dodatkowy atrybut
    
    def __str__(self):
        return f"Pies {self.imie} ({self.rasa}), {self.wiek} lat"

pies = Pies("Burek", 5, "Owczarek")
print(pies)
```

## 10. Enkapsulacja (ukrywanie danych)

```python
class KontoBankowe:
    def __init__(self, wlasciciel, saldo=0):
        self.wlasciciel = wlasciciel
        self.__saldo = saldo  # Atrybut prywatny (__ na początku)
    
    def wplata(self, kwota):
        if kwota > 0:
            self.__saldo += kwota
    
    def wyplata(self, kwota):
        if 0 < kwota <= self.__saldo:
            self.__saldo -= kwota
            return True
        return False
    
    def pobierz_saldo(self):
        """Getter - metoda dostępu do prywatnego atrybutu."""
        return self.__saldo

konto = KontoBankowe("Jan", 1000)
print(konto.wlasciciel)
# print(konto.__saldo)
print(f"Saldo: {konto.pobierz_saldo()} zł")

konto.wplata(500)
print(f"Po wpłacie: {konto.pobierz_saldo()} zł")

# Nie możemy bezpośrednio zmienić __saldo
# print(konto.__saldo)  # To spowoduje błąd!
```

## 11. Property - elegancki dostęp do atrybutów

```python
class Temperatura:
    def __init__(self, celsius=0):
        self._celsius = celsius
    
    @property
    def celsius(self):
        """Getter dla celsius."""
        return self._celsius
    
    @celsius.setter
    def celsius(self, wartosc):
        """Setter dla celsius z walidacją."""
        if wartosc < -273.15:
            raise ValueError("Temperatura nie może być niższa niż zero bezwzględne!")
        self._celsius = wartosc
    
    @property
    def fahrenheit(self):
        """Konwersja na Fahrenheita."""
        return self._celsius * 9/5 + 32
    
    @fahrenheit.setter
    def fahrenheit(self, wartosc):
        """Ustawienie przez Fahrenheita."""
        self._celsius = (wartosc - 32) * 5/9

# Użycie
temp = Temperatura(25)
print(temp._celsius)
print(f"Temperatura: {temp.celsius}°C = {temp.fahrenheit}°F")

temp.celsius = 30
print(f"Po zmianie: {temp.celsius}°C = {temp.fahrenheit}°F")

temp.fahrenheit = 100
print(f"Ustawiono 100°F: {temp.celsius:.2f}°C")
```

## 12. Metody klasowe i statyczne

```python
class Osoba:
    liczba_osob = 0  # Atrybut klasowy
    
    def __init__(self, imie, wiek):
        self.imie = imie
        self.wiek = wiek
        Osoba.liczba_osob += 1
    
    @classmethod
    def ile_osob(cls):
        """Metoda klasowa - dostęp do atrybutów klasy."""
        return f"Utworzono {cls.liczba_osob} osób"
    
    @staticmethod
    def czy_pelnoletni(wiek):
        """Metoda statyczna - nie potrzebuje dostępu do klasy ani instancji."""
        return wiek >= 18
    
    def __str__(self):
        return f"{self.imie}, {self.wiek} lat"

# Tworzenie obiektów
osoba1 = Osoba("Jan", 25)
osoba2 = Osoba("Anna", 17)
osoba3 = Osoba("Piotr", 30)

# Metoda klasowa
print(Osoba.ile_osob())

# Metoda statyczna
print(f"\nCzy Jan (25) pełnoletni? {Osoba.czy_pelnoletni(25)}")
print(f"Czy Anna (17) pełnoletnia? {Osoba.czy_pelnoletni(17)}")
```

## 13. Przykład kompleksowy: System biblioteki

```python
class Ksiazka:
    def __init__(self, tytul, autor, rok):
        self.tytul = tytul
        self.autor = autor
        self.rok = rok
        self.wypozyczona = False
    
    def __str__(self):
        status = "wypożyczona" if self.wypozyczona else "dostępna"
        return f"\"{self.tytul}\" - {self.autor} ({self.rok}) [{status}]"

class Biblioteka:
    def __init__(self, nazwa):
        self.nazwa = nazwa
        self.ksiazki = []
    
    def dodaj_ksiazke(self, ksiazka):
        self.ksiazki.append(ksiazka)
        print(f"Dodano: {ksiazka.tytul}")
    
    def wypozycz(self, tytul):
        for ksiazka in self.ksiazki:
            if ksiazka.tytul == tytul:
                if not ksiazka.wypozyczona:
                    ksiazka.wypozyczona = True
                    return f"Wypożyczono: {tytul}"
                else:
                    return f"Książka '{tytul}' jest już wypożyczona"
        return f"Nie znaleziono książki: {tytul}"
    
    def zwroc(self, tytul):
        for ksiazka in self.ksiazki:
            if ksiazka.tytul == tytul:
                if ksiazka.wypozyczona:
                    ksiazka.wypozyczona = False
                    return f"Zwrócono: {tytul}"
                else:
                    return f"Książka '{tytul}' nie była wypożyczona"
        return f"Nie znaleziono książki: {tytul}"
    
    def pokaz_ksiazki(self):
        print(f"\nKsiążki w bibliotece '{self.nazwa}':")
        for i, ksiazka in enumerate(self.ksiazki, 1):
            print(f"  {i}. {ksiazka}")

# Użycie systemu
biblioteka = Biblioteka("Miejska Biblioteka")

biblioteka.dodaj_ksiazke(Ksiazka("Pan Tadeusz", "Adam Mickiewicz", 1834))
biblioteka.dodaj_ksiazke(Ksiazka("Quo Vadis", "Henryk Sienkiewicz", 1896))
biblioteka.dodaj_ksiazke(Ksiazka("Lalka", "Bolesław Prus", 1890))

biblioteka.pokaz_ksiazki()

print(f"\n{biblioteka.wypozycz('Lalka')}")
print(f"{biblioteka.wypozycz('Lalka')}")

biblioteka.pokaz_ksiazki()

print(f"\n{biblioteka.zwroc('Lalka')}")
biblioteka.pokaz_ksiazki()
```
