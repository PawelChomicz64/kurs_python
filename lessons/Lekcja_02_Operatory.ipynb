## 1. Operatory arytmetyczne

Już poznałeś podstawowe operatory matematyczne. Zobaczmy je wszystkie razem:

```python
a = 15
b = 4

print(f"a = {a}, b = {b}")
print(f"a + b = {a + b}")  # Dodawanie
print(f"a - b = {a - b}")  # Odejmowanie
print(f"a * b = {a * b}")  # Mnożenie
print(f"a / b = {a / b}")  # Dzielenie (zawsze float)
print(f"a // b = {a // b}")  # Dzielenie całkowite
print(f"a % b = {a % b}")  # Reszta z dzielenia (modulo)
print(f"a ** b = {a ** b}")  # Potęgowanie
```

### Kolejność wykonywania operacji

Python przestrzega matematycznej kolejności działań (PEMDAS):
1. **P**arentheses (Nawiasy)
2. **E**xponents (Potęgowanie)
3. **M**ultiplication/**D**ivision (Mnożenie/Dzielenie) - od lewej do prawej
4. **A**ddition/**S**ubtraction (Dodawanie/Odejmowanie) - od lewej do prawej

```python
# Przykłady kolejności działań
wynik1 = 2 + 3 * 4
print(f"2 + 3 * 4 = {wynik1}")  # Najpierw mnożenie: 2 + 12 = 14

wynik2 = (2 + 3) * 4
print(f"(2 + 3) * 4 = {wynik2}")  # Najpierw nawiasy: 5 * 4 = 20

wynik3 = 2 ** 3 ** 2
print(f"2 ** 3 ** 2 = {wynik3}")  # Potęgowanie od prawej: 2 ** 9 = 512

wynik4 = 10 / 2 * 3
print(f"10 / 2 * 3 = {wynik4}")  # Od lewej do prawej: 5 * 3 = 15
```

## 2. Operatory przypisania

Operatory przypisania pozwalają na skrócony zapis operacji z przypisaniem:

```python
# Podstawowe przypisanie
x = 10
print(f"x = {x}")

# Przypisanie z dodawaniem
x += 5  # To samo co: x = x + 5
print(f"Po x += 5: x = {x}")

# Przypisanie z odejmowaniem
x -= 3  # To samo co: x = x - 3
print(f"Po x -= 3: x = {x}")

# Przypisanie z mnożeniem
x *= 2  # To samo co: x = x * 2
print(f"Po x *= 2: x = {x}")

# Przypisanie z dzieleniem
x /= 4  # To samo co: x = x / 4
print(f"Po x /= 4: x = {x}")

# Inne operatory przypisania
y = 17
y //= 5  # Dzielenie całkowite z przypisaniem
print(f"17 //= 5: y = {y}")

z = 17
z %= 5  # Modulo z przypisaniem
print(f"17 %= 5: z = {z}")

w = 2
w **= 3  # Potęgowanie z przypisaniem
print(f"2 **= 3: w = {w}")
```

## 3. Operatory porównania

Operatory porównania zwracają wartości logiczne (`True` lub `False`):

```python
a = 10
b = 5
c = 10

print(f"a = {a}, b = {b}, c = {c}")
print()

# Równość
print(f"a == b: {a == b}")  # False
print(f"a == c: {a == c}")  # True

# Nierówność
print(f"a != b: {a != b}")  # True
print(f"a != c: {a != c}")  # False

# Większe
print(f"a > b: {a > b}")   # True
print(f"b > a: {b > a}")   # False

# Mniejsze
print(f"a < b: {a < b}")   # False
print(f"b < a: {b < a}")   # True

# Większe lub równe
print(f"a >= c: {a >= c}")  # True
print(f"b >= a: {b >= a}")  # False

# Mniejsze lub równe
print(f"a <= c: {a <= c}")  # True
print(f"a <= b: {a <= b}")  # False
```

### Porównywanie stringów

Możemy też porównywać stringi!

```python
# Porównywanie stringów
imie1 = "Anna"
imie2 = "anna"
imie3 = "Anna"

print(f"imie1 == imie2: {imie1 == imie2}")  # False (wielkość liter ma znaczenie!)
print(f"imie1 == imie3: {imie1 == imie3}")  # True

# Porównanie bez uwzględnienia wielkości liter
print(f"imie1.lower() == imie2.lower(): {imie1.lower() == imie2.lower()}")  # True

# Porównywanie alfabetyczne
print(f"'apple' < 'banana': {'apple' < 'banana'}")  # True (alfabetycznie)
print(f"'Python' > 'Java': {'Python' > 'Java'}")    # True
```

## 4. Operatory logiczne

Operatory logiczne pozwalają łączyć warunki:
- `and` - prawda, gdy oba warunki są prawdziwe
- `or` - prawda, gdy przynajmniej jeden warunek jest prawdziwy
- `not` - odwraca wartość logiczną

```python
# Operator AND
print("=== Operator AND ===")
print(f"True and True = {True and True}")      # True
print(f"True and False = {True and False}")    # False
print(f"False and True = {False and True}")    # False
print(f"False and False = {False and False}")  # False

print("\n=== Operator OR ===")
# Operator OR
print(f"True or True = {True or True}")      # True
print(f"True or False = {True or False}")    # True
print(f"False or True = {False or True}")    # True
print(f"False or False = {False or False}")  # False

print("\n=== Operator NOT ===")
# Operator NOT
print(f"not True = {not True}")    # False
print(f"not False = {not False}")  # True
```

### Praktyczne przykłady operatorów logicznych

```python
wiek = 25
ma_prawo_jazdy = True
ma_samochod = False

# Sprawdzanie wielu warunków
moze_prowadzic = wiek >= 18 and ma_prawo_jazdy
print(f"Może prowadzić samochód: {moze_prowadzic}")

# Sprawdzanie alternatyw
moze_jezdzic = ma_samochod or ma_prawo_jazdy
print(f"Ma możliwość jazdy: {moze_jezdzic}")

# Negacja
potrzebuje_prawa_jazdy = not ma_prawo_jazdy
print(f"Potrzebuje prawa jazdy: {potrzebuje_prawa_jazdy}")

# Złożone warunki
temperatura = 22
jest_slonecznie = True
dobra_pogoda = (temperatura >= 20 and temperatura <= 30) and jest_slonecznie
print(f"Dobra pogoda na spacer: {dobra_pogoda}")
```

## 5. Operatory tożsamości

Operatory `is` i `is not` sprawdzają, czy dwie zmienne odnoszą się do tego samego obiektu w pamięci:

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a == b: {a == b}")  # True (mają tę samą wartość)
print(f"a is b: {a is b}")  # False (to różne obiekty w pamięci)
print(f"a is c: {a is c}")  # True (c wskazuje na ten sam obiekt co a)

# Dla liczb i stringów Python czasem optymalizuje
x = 5
y = 5
print(f"x is y: {x is y}")  # True (Python używa tego samego obiektu dla małych liczb)

# Operator is not
print(f"a is not b: {a is not b}")  # True
```

## 6. Operatory przynależności

Operatory `in` i `not in` sprawdzają, czy element znajduje się w sekwencji:

```python
# Sprawdzanie w stringach
tekst = "Python jest super"
print(f"'Python' in tekst: {'Python' in tekst}")  # True
print(f"'Java' in tekst: {'Java' in tekst}")      # False
print(f"'Java' not in tekst: {'Java' not in tekst}")  # True

# Sprawdzanie w listach (poznasz je dokładniej w następnej lekcji)
owoce = ["jabłko", "banan", "pomarańcza"]
print(f"'banan' in owoce: {'banan' in owoce}")  # True
print(f"'gruszka' in owoce: {'gruszka' in owoce}")  # False

# Sprawdzanie cyfr
liczby = "12459"
print(f"'3' in liczby: {'3' in liczby}")  # True
print(f"'9' in liczby: {'9' in liczby}")  # False
```

## 7. Konwersje typów

Czasami musimy przekonwertować dane z jednego typu na inny:

```python
# String -> Integer
tekst_liczba = "42"
liczba = int(tekst_liczba)
print(f"String '42' -> int: {liczba}, typ: {type(liczba)}")

# String -> Float
tekst_float = "3.14"
liczba_float = float(tekst_float)
print(f"String '3.14' -> float: {liczba_float}, typ: {type(liczba_float)}")

# Integer -> String
liczba_int = 100
tekst = str(liczba_int)
print(f"Int 100 -> string: '{tekst}', typ: {type(tekst)}")

# Float -> Integer (ucina część dziesiętną!)
pi = 3.14159
pi_int = int(pi)
print(f"Float 3.14159 -> int: {pi_int}")

# Integer -> Float
calkowita = 5
zmiennoprzecinkowa = float(calkowita)
print(f"Int 5 -> float: {zmiennoprzecinkowa}")

# Boolean -> Integer
print(f"int(True) = {int(True)}")   # 1
print(f"int(False) = {int(False)}")  # 0

# Integer -> Boolean
print(f"bool(0) = {bool(0)}")    # False
print(f"bool(1) = {bool(1)}")    # True
print(f"bool(42) = {bool(42)}")  # True (każda liczba różna od 0 to True)
```
