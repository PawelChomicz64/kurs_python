# Lekcja 6: Funkcje

Funkcje to bloki kodu, które możemy wielokrotnie używać. Pozwalają na organizację kodu i unikanie powtórzeń.

## 1. Definiowanie funkcji

Funkcję definiujemy używając słowa kluczowego `def`.

```python
# Najprostsza funkcja
def powitanie():
    print("Witaj w świecie Pythona!")

# Wywołanie funkcji
powitanie()
powitanie()  # Możemy wywołać wielokrotnie
```

## 2. Funkcje z parametrami

Parametry pozwalają przekazywać dane do funkcji.

```python
# Funkcja z jednym parametrem
def powitaj_osobe(imie):
    print(f"Witaj, {imie}!")

powitaj_osobe("Anna")
powitaj_osobe("Jan")

# Funkcja z wieloma parametrami
def przedstaw_sie(imie, wiek, miasto):
    print(f"Cześć, mam na imię {imie}, mam {wiek} lat i mieszkam w {miasto}.")

przedstaw_sie("Maria", 25, "Warszawa")
przedstaw_sie("Piotr", 30, "Kraków")
```

## 3. Zwracanie wartości

Funkcje mogą zwracać wyniki używając słowa kluczowego `return`.

```python
# Funkcja zwracająca wartość
def dodaj(a, b):
    return a + b

wynik = dodaj(5, 3)
print(f"5 + 3 = {wynik}")

# Możemy używać wyniku bezpośrednio
print(f"10 + 20 = {dodaj(10, 20)}")

# Funkcja obliczająca pole prostokąta
def pole_prostokata(dlugosc, szerokosc):
    return dlugosc * szerokosc

pole = pole_prostokata(5, 3)
print(f"\nPole prostokąta 5x3: {pole}")
```

## 4. Zwracanie wielu wartości

```python
# Funkcja zwracająca wiele wartości (krotkę)
def statystyki(lista):
    minimum = min(lista)
    maksimum = max(lista)
    srednia = sum(lista) / len(lista)
    return minimum, maksimum, srednia

liczby = [3, 1, 4, 1, 5, 9, 2, 6]
min_val, max_val, avg = statystyki(liczby)

print(f"Minimum: {min_val}")
print(f"Maksimum: {max_val}")
print(f"Średnia: {avg:.2f}")

# Możemy też otrzymać jako krotkę
wyniki = statystyki(liczby)
print(f"\nWyniki jako krotka: {wyniki}")
```

## 5. Parametry domyślne

Możemy określić domyślne wartości parametrów.

```python
def powitanie(imie, jezyk="polski"):
    if jezyk == "polski":
        print(f"Cześć, {imie}!")
    elif jezyk == "angielski":
        print(f"Hello, {imie}!")
    elif jezyk == "hiszpanski":
        print(f"¡Hola, {imie}!")

# Używamy domyślnej wartości
powitanie("Anna")

# Podajemy własną wartość
powitanie("John", "angielski")
powitanie("Maria", "hiszpanski")

# Przykład z wieloma parametrami domyślnymi
def potega(podstawa, wykladnik=2):
    return podstawa ** wykladnik

print(f"\n5^2 = {potega(5)}")
print(f"5^3 = {potega(5, 3)}")
print(f"2^10 = {potega(2, 10)}")
```

## 6. Argumenty nazwane (keyword arguments)

```python
def opisz_zwierze(nazwa, typ, wiek):
    print(f"{nazwa} to {typ}, ma {wiek} lat.")

# Argumenty pozycyjne
opisz_zwierze("Burek", "pies", 5)

# Argumenty nazwane - kolejność nie ma znaczenia
opisz_zwierze(nazwa="Mruczek", wiek=3, typ="kot")
opisz_zwierze(wiek=2, nazwa="Tweety", typ="papuga")

# Mieszane (pozycyjne muszą być przed nazwanymi)
opisz_zwierze("Reksio", typ="pies", wiek=7)
```

## 7. *args - zmienna liczba argumentów pozycyjnych

```python
# *args pozwala przekazać dowolną liczbę argumentów
def suma(*liczby):
    wynik = 0
    for liczba in liczby:
        wynik += liczba
    return wynik

print(f"suma(1, 2, 3) = {suma(1, 2, 3)}")
print(f"suma(1, 2, 3, 4, 5) = {suma(1, 2, 3, 4, 5)}")
print(f"suma(10, 20) = {suma(10, 20)}")

# Przykład z dodatkowymi parametrami
def przedstaw_liste(tytul, *elementy):
    print(f"\n{tytul}:")
    for element in elementy:
        print(f"  - {element}")

przedstaw_liste("Moje hobby", "programowanie", "czytanie", "sport")
przedstaw_liste("Ulubione owoce", "jabłko", "banan")
```

## 8. **kwargs - zmienna liczba argumentów nazwanych

```python
# **kwargs pozwala przekazać dowolną liczbę argumentów nazwanych
def wyswietl_info(**dane):
    for klucz, wartosc in dane.items():
        print(f"{klucz}: {wartosc}")

print("Osoba 1:")
wyswietl_info(imie="Jan", wiek=30, miasto="Warszawa")

print("\nOsoba 2:")
wyswietl_info(imie="Anna", zawod="programista", email="anna@example.com")

# Łączenie *args i **kwargs
def pelna_funkcja(obowiazkowy, *args, domyslny="wartość", **kwargs):
    print(f"Obowiązkowy: {obowiazkowy}")
    print(f"Args: {args}")
    print(f"Domyślny: {domyslny}")
    print(f"Kwargs: {kwargs}")

print("\nPełna funkcja:")
pelna_funkcja(1, 2, 3, klucz1="wartość1", klucz2="wartość2", k=3, obowiazkowy="test")
```

## 9. Funkcje lambda

Funkcje lambda to małe, anonimowe funkcje w jednej linii.

```python
# Tradycyjna funkcja
def kwadrat(x):
    return x ** 2

# Funkcja lambda
kwadrat_lambda = lambda x: x ** 2

print(f"Tradycyjna: {kwadrat(5)}")
print(f"Lambda: {kwadrat_lambda(5)}")

# Lambda z wieloma argumentami
suma = lambda a, b: a + b
print(f"\nSuma: {suma(3, 7)}")

# Lambda w sortowaniu
punkty = [(1, 2), (3, 1), (5, 4), (2, 3)]
print(f"\nOryginalne punkty: {punkty}")

# Sortowanie po pierwszej współrzędnej
sortowane_x = sorted(punkty, key=lambda punkt: punkt[0])
print(f"Sortowane po x: {sortowane_x}")

# Sortowanie po drugiej współrzędnej
sortowane_y = sorted(punkty, key=lambda punkt: punkt[1])
print(f"Sortowane po y: {sortowane_y}")

# Lambda z map()
liczby = [1, 2, 3, 4, 5]
kwadraty = list(map(lambda x: x**2, liczby))
print(f"\nKwadraty: {kwadraty}")

# Lambda z filter()
parzyste = list(filter(lambda x: x % 2 == 0, liczby))
print(f"Parzyste: {parzyste}")
```

## 10. Zakres zmiennych (scope)

Zmienne mają różny zakres widoczności.

```python
# Zmienna globalna
x = 10

def funkcja1():
    # Zmienna lokalna
    y = 5
    print(f"Wewnątrz funkcji1: x = {x}, y = {y}")

funkcja1()
print(f"Na zewnątrz: x = {x}")
# print(y)  # To spowoduje błąd - y nie istnieje poza funkcją

# Modyfikowanie zmiennej globalnej
licznik = 0

def zwieksz_licznik():
    global licznik  # Deklarujemy, że używamy zmiennej globalnej
    licznik += 1

print(f"\nLicznik przed: {licznik}")
zwieksz_licznik()
print(f"Licznik po: {licznik}")
zwieksz_licznik()
print(f"Licznik po: {licznik}")
```

## 11. Dokumentacja funkcji (docstrings)

```python
def oblicz_bmi(waga, wzrost):
    """
    Oblicza wskaźnik masy ciała (BMI).
    
    Parametry:
        waga (float): Waga w kilogramach
        wzrost (float): Wzrost w metrach
    
    Zwraca:
        float: Wartość BMI
    
    Przykład:
        >>> oblicz_bmi(70, 1.75)
        22.86
    """
    return waga / (wzrost ** 2)

# Wywołanie funkcji
bmi = oblicz_bmi(70, 1.75)
print(f"BMI: {bmi:.2f}")

# Wyświetlenie dokumentacji
print("\nDokumentacja funkcji:")
print(oblicz_bmi.__doc__)

# W Jupyter możesz też użyć:
# help(oblicz_bmi)
# lub oblicz_bmi?
```

## 12. Funkcje rekurencyjne

Funkcja może wywoływać samą siebie.

```python
# Silnia (factorial)
def silnia(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * silnia(n - 1)

print("Silnie:")
for i in range(6):
    print(f"{i}! = {silnia(i)}")

# Ciąg Fibonacciego
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print("\nCiąg Fibonacciego:")
for i in range(10):
    print(f"F({i}) = {fibonacci(i)}")
```

## 13. Funkcje wyższego rzędu

Funkcje mogą przyjmować inne funkcje jako argumenty.

```python
def zastosuj_operacje(lista, operacja):
    """
    Stosuje operację do każdego elementu listy.
    """
    return [operacja(x) for x in lista]

def kwadrat(x):
    return x ** 2

def podwoj(x):
    return x * 2

liczby = [1, 2, 3, 4, 5]

print(f"Oryginalna lista: {liczby}")
print(f"Kwadraty: {zastosuj_operacje(liczby, kwadrat)}")
print(f"Podwojone: {zastosuj_operacje(liczby, podwoj)}")

# Z funkcją lambda
print(f"Sześciany: {zastosuj_operacje(liczby, lambda x: x**3)}")
```
