## 1. Instrukcja if

Instrukcja `if` pozwala wykonać kod tylko wtedy, gdy warunek jest prawdziwy.

**Ważne:** W Pythonie wcięcia (indentacja) są obowiązkowe! Standardowo używamy 4 spacji.

````python
# Podstawowa instrukcja if
wiek = 15

if wiek >= 18:
    print("Jesteś pełnoletni!")
    print("Możesz głosować.")

print("To zdanie wykona się zawsze.")
````

````python
# Przykład z fałszywym warunkiem
temperatura = 15

if temperatura > 25:
    print("Jest gorąco!")
    print("Idę na plażę.")

print("Sprawdzam pogodę.")
````

## 2. Instrukcja if-else

Instrukcja `else` wykonuje się, gdy warunek w `if` jest fałszywy.

````python
wiek = 16

if wiek >= 18:
    print("Jesteś pełnoletni.")
else:
    print("Jesteś niepełnoletni.")
    lata_do_pelnoletnosci = 18 - wiek
    print(f"Do pełnoletności zostało Ci {lata_do_pelnoletnosci} lat.")
````

````python
# Praktyczny przykład: sprawdzanie parzystości
liczba = 7

if liczba % 2 == 0:
    print(f"{liczba} jest liczbą parzystą.")
else:
    print(f"{liczba} jest liczbą nieparzystą.")
````

## 3. Instrukcja if-elif-else

`elif` (else if) pozwala sprawdzić wiele warunków po kolei.

````python
ocena = 85

if ocena >= 90:
    print("Ocena: Celujący (5+)")
elif ocena >= 80:
    print("Ocena: Bardzo dobry (5)")
elif ocena >= 70:
    print("Ocena: Dobry (4)")
elif ocena >= 60:
    print("Ocena: Dostateczny (3)")
elif ocena >= 50:
    print("Ocena: Dopuszczający (2)")
else:
    print("Ocena: Niedostateczny (1)")
````

````python
# Przykład: klasyfikacja BMI
waga = 70  # kg
wzrost = 1.75  # m
bmi = waga / (wzrost ** 2)

print(f"Twoje BMI: {bmi:.2f}")

if bmi < 18.5:
    print("Niedowaga")
elif bmi < 25:
    print("Waga prawidłowa")
elif bmi < 30:
    print("Nadwaga")
else:
    print("Otyłość")
````

## 4. Zagnieżdżone instrukcje if

Możesz umieszczać instrukcje `if` wewnątrz innych instrukcji `if`.

````python
wiek = 25
ma_prawo_jazdy = True

if wiek >= 18:
    print("Jesteś pełnoletni.")
    if ma_prawo_jazdy:
        print("Możesz prowadzić samochód.")
    else:
        print("Musisz zdobyć prawo jazdy.")
else:
    print("Jesteś za młody, aby prowadzić samochód.")
````

## 5. Operator warunkowy (ternary operator)

Python ma skróconą formę instrukcji if-else w jednej linii:

````python
# Tradycyjna forma
wiek = 20
if wiek >= 18:
    status = "pełnoletni"
else:
    status = "niepełnoletni"
print(f"Status (tradycyjnie): {status}")

# Operator warunkowy (ternary)
status = "pełnoletni" if wiek >= 18 else "niepełnoletni"
print(f"Status (ternary): {status}")

# Więcej przykładów
liczba = 15
typ = "parzysta" if liczba % 2 == 0 else "nieparzysta"
print(f"{liczba} jest {typ}")

temperatura = 22
pogoda = "ciepło" if temperatura > 20 else "zimno"
print(f"Jest {pogoda}")
````

## 6. Pętla while

Pętla `while` wykonuje kod, dopóki warunek jest prawdziwy.

````python
# Podstawowa pętla while
licznik = 1

while licznik <= 5:
    print(f"Licznik: {licznik}")
    licznik += 1  # Ważne! Musimy zwiększać licznik, inaczej pętla będzie nieskończona

print("Pętla zakończona!")
````

````python
# Przykład: suma liczb od 1 do 10
suma = 0
liczba = 1

while liczba <= 10:
    suma += liczba
    liczba += 1

print(f"Suma liczb od 1 do 10: {suma}")
````

````python
# Przykład: zgadywanie liczby
sekretna_liczba = 7
proba = 0

# Symulacja zgadywania (w prawdziwym programie użylibyśmy input())
zgadniecia = [3, 5, 7]  # Symulowane odpowiedzi użytkownika
indeks = 0

while indeks < len(zgadniecia):
    zgadniecie = zgadniecia[indeks]
    proba += 1
    
    if zgadniecie == sekretna_liczba:
        print(f"Gratulacje! Zgadłeś za {proba} razem!")
        break  # Przerywa pętlę
    elif zgadniecie < sekretna_liczba:
        print(f"Próba {proba}: {zgadniecie} - Za mało!")
    else:
        print(f"Próba {proba}: {zgadniecie} - Za dużo!")
    
    indeks += 1
````

## 7. Pętla for

Pętla `for` iteruje po sekwencji (liście, stringu, zakresie liczb, itp.).

````python
# Iteracja po zakresie liczb
print("Liczby od 0 do 4:")

for i in range(5):  # range(5) generuje: 0, 1, 2, 3, 4
    print(i)

print("\nLiczby od 1 do 5:")
for i in range(1, 6):  # range(start, stop)
    print(i)

print("\nCo druga liczba od 0 do 10:")
for i in range(0, 11, 2):  # range(start, stop, step)
    print(i)
````

````python
# Iteracja po stringu
slowo = "12345"

for litera in slowo:
    print(f"Litera: {litera}")
    print(type(litera))
````

````python
# Iteracja po liście
owoce = ["jabłko", "banan", "pomarańcza", "gruszka"]

for owoc in owoce:
    print(f"Lubię {owoc}")
````

````python
# Przykład: tabliczka mnożenia
liczba = 7

print(f"Tabliczka mnożenia dla {liczba}:")
for i in range(1, 11):
    wynik = liczba * i
    print(f"{liczba} × {i} = {wynik}")
````

````python
# Funkcja enumerate() - iteracja z indeksem
kolory = ["czerwony", "zielony", "niebieski"]

for indeks, kolor in enumerate(kolory):
    print(f"Kolor {indeks}: {kolor}")

# Możemy zacząć od innego numeru
print("\nZ numeracją od 1:")
for indeks, kolor in enumerate(kolory[-2:], start=0):
    print(f"Kolor {indeks}: {kolor}")
````

## 8. Instrukcje break, continue i pass

Te instrukcje pozwalają kontrolować przepływ pętli.

### break - przerywa pętlę

````python
# Szukanie pierwszej liczby podzielnej przez 7
for liczba in range(1, 100):
    if liczba % 7 == 0:
        print(f"Pierwsza liczba podzielna przez 7: {liczba}")
        break  # Przerywa pętlę

# Przykład z while
print("\nLiczenie do nieskończoności (z przerwaniem):")
licznik = 1
while True:  # Nieskończona pętla
    print(licznik)
    licznik += 1
    if licznik > 5:
        break  # Przerywa nieskończoną pętlę
````

### continue - pomija resztę iteracji i przechodzi do następnej

````python
# Wyświetl tylko liczby nieparzyste
print("Liczby nieparzyste od 1 do 10:")
for liczba in range(1, 11):
    if liczba % 2 == 0:  # Jeśli parzysta
        continue  # Pomiń resztę i przejdź do następnej iteracji
    print(liczba)

# Pomijanie określonych wartości
print("\nLiczby od 1 do 10 (bez 5):")
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
````

### pass - nie robi nic (placeholder)

````python
# pass jest używany jako placeholder
for i in range(5):
    if i == 2:
        pass  # TODO: Zaimplementuj coś później
    else:
        print(i)

# Przydatne przy tworzeniu struktury kodu
wiek = 25
if wiek < 18:
    pass  # Zaimplementuję to później
elif wiek < 65:
    print("Wiek produkcyjny")
else:
    pass  # Zaimplementuję to później
````

## 9. Zagnieżdżone pętle

Możesz umieszczać pętle wewnątrz innych pętli.

````python
# Przykład: wyświetlanie prostokąta z gwiazdek
wysokosc = 3
szerokosc = 5

for wiersz in range(wysokosc):
    for kolumna in range(szerokosc):
        print("*", end="")  # end="" zapobiega przejściu do nowej linii
    print()  # Nowa linia po każdym wierszu
````

````python
# Przykład: tabliczka mnożenia (pełna)
print("Tabliczka mnożenia 1-5:")
print()

for i in range(1, 6):
    for j in range(1, 6):
        wynik = i * j
        print(f"{wynik:3}", end=" ")  # :3 wyrównuje do 3 znaków
    print()  # Nowa linia po każdym wierszu
````

````python
# Przykład: trójkąt z liczb
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

## 10. Pętla for z else

Python ma unikalną cechę: blok `else` w pętli wykonuje się, gdy pętla zakończy się normalnie (bez `break`).

````python
# Szukanie liczby pierwszej
liczba = 29

for dzielnik in range(2, liczba):
    if liczba % dzielnik == 0:
        print(f"{liczba} nie jest liczbą pierwszą (dzieli się przez {dzielnik})")
        break
else:
    # Ten blok wykona się tylko jeśli pętla NIE została przerwana przez break
    print(f"{liczba} jest liczbą pierwszą!")
````

````python
# Przykład z przerwaniem
liczba = 15

for dzielnik in range(2, liczba):
    if liczba % dzielnik == 0:
        print(f"{liczba} nie jest liczbą pierwszą (dzieli się przez {dzielnik})")
        break
else:
    print(f"{liczba} jest liczbą pierwszą!")
````
