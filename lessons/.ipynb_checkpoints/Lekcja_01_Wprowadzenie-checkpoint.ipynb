## 2. Zmienne

Zmienne to "pojemniki" na dane. W Pythonie nie musisz deklarować typu zmiennej - Python sam to rozpozna!

````python
# To jest komentarz - Python go ignoruje
# Komentarze zaczynają się od znaku #

# Tworzenie zmiennych
imie = "Anna"
wiek = 25
wzrost = 1.68
student = True

# Wyświetlanie zmiennych
print("Imię:", imie)
print("Wiek:", wiek)
print("Wzrost:", wzrost, "m")
print("Czy student:", student)
````

### Zasady nazewnictwa zmiennych:
- Mogą zawierać litery, cyfry i podkreślenia
- Muszą zaczynać się od litery lub podkreślenia
- Są wrażliwe na wielkość liter (`wiek` ≠ `Wiek`)
- Nie mogą być słowami kluczowymi Pythona (np. `print`, `if`, `for`)

**Dobre nazwy:**
```python
imie_uzytkownika
liczba_studentow
suma_total
```

**Złe nazwy:**
```python
1zmienna  # zaczyna się od cyfry
moja-zmienna  # zawiera myślnik
print  # słowo kluczowe
```

## 3. Typy danych

Python ma kilka podstawowych typów danych:

````python
# Integer (int) - liczby całkowite
liczba_calkowita = 42
print("Integer:", liczba_calkowita, "- typ:", type(liczba_calkowita))

# Float - liczby zmiennoprzecinkowe
liczba_zmiennoprzecinkowa = 3.14
print("Float:", liczba_zmiennoprzecinkowa, "- typ:", type(liczba_zmiennoprzecinkowa))

# String (str) - tekst
tekst = "Python jest super!"
print("String:", tekst, "- typ:", type(tekst))

# Boolean (bool) - wartości logiczne
prawda = True
falsz = False
print("Boolean:", prawda, "- typ:", type(prawda))
````

Funkcja `type()` pozwala sprawdzić typ zmiennej.

## 4. Operacje na stringach

Stringi (teksty) to bardzo ważny typ danych. Możemy z nimi robić wiele ciekawych rzeczy!

````python
# Łączenie stringów (konkatenacja)
imie = "Jan"
nazwisko = "Kowalski"
pelne_imie = imie + " " + nazwisko
print("Pełne imię:", pelne_imie)

# Powtarzanie stringów
smiechy = "ha" * 5
print(smiechy)

# Długość stringa
tekst = "Python"
print("Długość tekstu:", len(tekst))

# Indeksowanie (numeracja od 0!)
print("Pierwsza litera:", tekst[0])
print("Ostatnia litera:", tekst[-1])

# Wycinanie (slicing)
print("Pierwsze 3 litery:", tekst[0:3])
print("Od 2 do końca:", tekst[2:])
````

## 5. Formatowanie stringów

Python oferuje kilka sposobów na wstawianie zmiennych do tekstów:

````python
imie = "Maria"
wiek = 30
wzrost = 1.65

# Sposób 1: f-string (najnowszy i najlepszy!)
wiadomosc = f"Cześć, mam na imię {imie}, mam {wiek} lat i {wzrost}m wzrostu."
print(wiadomosc)

# Możemy też wykonywać operacje wewnątrz f-string
print(f"Za rok będę mieć {wiek + 1} lat.")
print(f"Mój wzrost w cm: {wzrost * 100}cm")

# Sposób 2: metoda .format()
wiadomosc2 = "Cześć, mam na imię {}, mam {} lat.".format(imie, wiek)
print(wiadomosc2)

# Sposób 3: konkatenacja (najmniej wygodny)
wiadomosc3 = "Cześć, mam na imię " + imie + ", mam " + str(wiek) + " lat."
print(wiadomosc3)
````

## 6. Pobieranie danych od użytkownika

Funkcja `input()` pozwala pobrać dane od użytkownika:

````python
# Uwaga: input() zawsze zwraca string!
imie_uzytkownika = input("Jak masz na imię? ")
print(f"Miło Cię poznać, {imie_uzytkownika}!")

# Jeśli potrzebujemy liczby, musimy przekonwertować
wiek_tekst = input("Ile masz lat? ")
wiek_liczba = int(wiek_tekst)  # Konwersja string -> int
print(f"Za 10 lat będziesz mieć {wiek_liczba + 10} lat!")
````

## 7. Podstawowe operacje matematyczne

````python
a = 10
b = 3

print("Dodawanie:", a + b)
print("Odejmowanie:", a - b)
print("Mnożenie:", a * b)
print("Dzielenie:", a / b)  # Zawsze zwraca float
print("Dzielenie całkowite:", a // b)  # Zwraca int
print("Reszta z dzielenia (modulo):", a % b)
print("Potęgowanie:", a ** b)  # 10 do potęgi 3
````
