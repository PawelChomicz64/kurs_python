## 1. Otwieranie i zamykanie plików

## 2. Zapis do pliku

```python
# Sposób 1: Ręczne zamykanie
plik = open('/home/ubuntu/python_course/resources/test1.txt', 'w')
plik.write("Witaj, świecie!\n")
plik.write("To jest druga linia.\n")
plik.close()  # Ważne: zawsze zamykaj plik!

print("Zapisano plik test1.txt")
```

```python
# Sposób 2: Context manager (with) - ZALECANY!
with open('/home/ubuntu/python_course/resources/test2.txt', 'w') as plik:
    plik.write("Pierwsza linia\n")
    plik.write("Druga linia\n")
    plik.write("Trzecia linia\n")
# Plik automatycznie się zamyka po wyjściu z bloku with

print("Zapisano plik test2.txt")
```

## 5. Praca z plikami CSV

```python
import csv

# Zapisywanie do CSV
dane = [
    ['Imię', 'Wiek', 'Miasto'],
    ['Jan', '30', 'Warszawa'],
    ['Anna', '25', 'Kraków'],
    ['Piotr', '35', 'Gdańsk']
]

with open('/home/ubuntu/python_course/resources/osoby.csv', 'w', newline='') as plik:
    writer = csv.writer(plik)
    writer.writerows(dane)

print("Zapisano plik CSV")

# Odczyt z CSV
with open('/home/ubuntu/python_course/resources/osoby.csv', 'r') as plik:
    reader = csv.reader(plik)
    print("\nZawartość CSV:")
    for wiersz in reader:
        print(wiersz)
```

```python
# CSV ze słownikami
dane_dict = [
    {'imie': 'Maria', 'wiek': 28, 'miasto': 'Poznań'},
    {'imie': 'Tomek', 'wiek': 32, 'miasto': 'Wrocław'}
]

with open('/home/ubuntu/python_course/resources/osoby_dict.csv', 'w', newline='') as plik:
    fieldnames = ['imie', 'wiek', 'miasto']
    writer = csv.DictWriter(plik, fieldnames=fieldnames)
    
    writer.writeheader()  # Zapisz nagłówki
    writer.writerows(dane_dict)

# Odczyt
with open('/home/ubuntu/python_course/resources/osoby_dict.csv', 'r') as plik:
    reader = csv.DictReader(plik)
    print("\nOdczyt jako słowniki:")
    for wiersz in reader:
        print(f"{wiersz['imie']} ({wiersz['wiek']}) - {wiersz['miasto']}")
```

## 6. Praca z plikami JSON

```python
import json

# Dane do zapisania
dane = {
    'imie': 'Jan',
    'wiek': 30,
    'miasto': 'Warszawa',
    'hobby': ['programowanie', 'czytanie', 'sport'],
    'aktywny': True
}

# Zapis do JSON
with open('/home/ubuntu/python_course/resources/dane.json', 'w', encoding='utf-8') as plik:
    json.dump(dane, plik, indent=2, ensure_ascii=False)

print("Zapisano JSON")

# Odczyt z JSON
with open('/home/ubuntu/python_course/resources/dane.json', 'r', encoding='utf-8') as plik:
    wczytane_dane = json.load(plik)
    print("\nWczytane dane:")
    print(wczytane_dane)
    print(f"\nImię: {wczytane_dane['imie']}")
    print(f"Hobby: {', '.join(wczytane_dane['hobby'])}")
```

## 7. Obsługa wyjątków (try-except)

Wyjątki pozwalają obsługiwać błędy w programie.

```python
# Bez obsługi wyjątków - program się zatrzyma
# liczba = int("abc")  # To spowoduje ValueError!

# Z obsługą wyjątków
try:
    liczba = int("abc")
    print(f"Liczba: {liczba}")
except ValueError:
    print("Błąd: Nie można przekonwertować na liczbę!")

print("Program działa dalej...")
```

```python
# Obsługa wielu typów wyjątków
def dziel(a, b):
    try:
        wynik = a / b
        return wynik
    except ZeroDivisionError:
        print("Błąd: Nie można dzielić przez zero!")
        return None
    except TypeError:
        print("Błąd: Nieprawidłowy typ danych!")
        return None

print(f"10 / 2 = {dziel(10, 2)}")
print(f"10 / 0 = {dziel(10, 0)}")
print(f"10 / 'a' = {dziel(10, 'a')}")
```

```python
# try-except-else-finally
def czytaj_plik(nazwa_pliku):
    try:
        plik = open(nazwa_pliku, 'r')
        zawartosc = plik.read()
    except FileNotFoundError:
        print(f"Błąd: Plik '{nazwa_pliku}' nie istnieje!")
        zawartosc = None
    except PermissionError:
        print(f"Błąd: Brak uprawnień do pliku '{nazwa_pliku}'!")
        zawartosc = None
    else:
        print("Plik odczytany pomyślnie!")
    finally:
        print("Blok finally wykonuje się zawsze!")
        try:
            plik.close()
        except:
            pass
    
    return zawartosc

# Testowanie
print("Test 1: Istniejący plik")
wynik = czytaj_plik('/home/ubuntu/python_course/resources/test2.txt')

print("\nTest 2: Nieistniejący plik")
wynik = czytaj_plik('/home/ubuntu/python_course/resources/nieistnieje.txt')
```

## 8. Przechwytywanie informacji o błędzie

```python
try:
    wynik = 10 / 0
except ZeroDivisionError as e:
    print(f"Wystąpił błąd: {e}")
    print(f"Typ błędu: {type(e).__name__}")

# Przechwytywanie dowolnego wyjątku
try:
    lista = [1, 2, 3]
    print(lista[10])
except Exception as e:
    print(f"\nWystąpił błąd: {e}")
    print(f"Typ: {type(e).__name__}")
```

## 9. Wywoływanie własnych wyjątków

```python
def sprawdz_wiek(wiek):
    if wiek < 0:
        raise ValueError("Wiek nie może być ujemny!")
    if wiek > 150:
        raise ValueError("Wiek wydaje się nieprawdopodobny!")
    return f"Wiek {wiek} jest prawidłowy"

# Testowanie
try:
    print(sprawdz_wiek(25))
    print(sprawdz_wiek(-5))
except ValueError as e:
    print(f"Błąd walidacji: {e}")
```

## 10. Własne klasy wyjątków

```python
class NiewystarczajaceSrodkiError(Exception):
    """Wyjątek wywoływany gdy brakuje środków na koncie."""
    pass

class KontoBankowe:
    def __init__(self, saldo=0):
        self.saldo = saldo
    
    def wyplata(self, kwota):
        if kwota > self.saldo:
            raise NiewystarczajaceSrodkiError(
                f"Niewystarczające środki! Saldo: {self.saldo}, próba wypłaty: {kwota}"
            )
        self.saldo -= kwota
        return f"Wypłacono {kwota} zł. Nowe saldo: {self.saldo} zł"

# Użycie
konto = KontoBankowe(1000)

try:
    print(konto.wyplata(500))
    print(konto.wyplata(700))  # Za dużo!
except NiewystarczajaceSrodkiError as e:
    print(f"Błąd: {e}")
```

## 11. Praktyczny przykład: System logowania

```python
import datetime

class Logger:
    def __init__(self, nazwa_pliku):
        self.nazwa_pliku = nazwa_pliku
    
    def log(self, wiadomosc, poziom="INFO"):
        """Zapisuje wiadomość do pliku logu."""
        timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        linia = f"[{timestamp}] [{poziom}] {wiadomosc}\n"
        
        try:
            with open(self.nazwa_pliku, 'a', encoding='utf-8') as plik:
                plik.write(linia)
        except Exception as e:
            print(f"Błąd zapisu do logu: {e}")
    
    def info(self, wiadomosc):
        self.log(wiadomosc, "INFO")
    
    def warning(self, wiadomosc):
        self.log(wiadomosc, "WARNING")
    
    def error(self, wiadomosc):
        self.log(wiadomosc, "ERROR")
    
    def pokaz_logi(self):
        """Wyświetla zawartość pliku logu."""
        try:
            with open(self.nazwa_pliku, 'r', encoding='utf-8') as plik:
                print(plik.read())
        except FileNotFoundError:
            print("Plik logu nie istnieje.")

# Użycie
logger = Logger('/home/ubuntu/python_course/resources/app.log')

logger.info("Aplikacja uruchomiona")
logger.info("Użytkownik zalogowany: Jan")
logger.warning("Niska ilość pamięci")
logger.error("Błąd połączenia z bazą danych")
logger.info("Aplikacja zamknięta")

print("Zawartość logu:")
print("="*50)
logger.pokaz_logi()
```
