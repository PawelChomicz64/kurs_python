## 2. Importowanie modułów

Python ma wbudowaną bogatą bibliotekę standardową.

## 3. Moduł random - liczby losowe

```python
import random

# Losowa liczba zmiennoprzecinkowa między 0 a 1
print(f"random(): {random.random()}")

# Losowa liczba całkowita w zakresie
print(f"\nrandint(1, 10): {random.randint(1, 10)}")
print(f"randint(1, 100): {random.randint(1, 100)}")

# Losowy wybór z listy
kolory = ["czerwony", "zielony", "niebieski", "żółty"]
print(f"\nchoice(kolory): {random.choice(kolory)}")

# Losowe mieszanie listy
karty = ["As", "Król", "Dama", "Walet", "10"]
print(f"\nPrzed shuffle: {karty}")
random.shuffle(karty)
print(f"Po shuffle: {karty}")

# Losowa próbka (sample)
liczby = list(range(1, 50))
losowe_6 = random.sample(liczby, 6)  # Losuj 6 unikalnych liczb
print(f"\nLosowe 6 liczb z 49: {sorted(losowe_6)}")
```

## 4. Moduł datetime - daty i czas

```python
from datetime import datetime, date, time, timedelta

# Aktualna data i czas
teraz = datetime.now()
print(f"Teraz: {teraz}")
print(f"Rok: {teraz.year}")
print(f"Miesiąc: {teraz.month}")
print(f"Dzień: {teraz.day}")
print(f"Godzina: {teraz.hour}")
print(f"Minuta: {teraz.minute}")

# Dzisiejsza data
dzis = date.today()
print(f"\nDzisiaj: {dzis}")

# Tworzenie konkretnej daty
urodziny = date(1990, 5, 15)
print(f"Urodziny: {urodziny}")

# Różnica dat
roznica = dzis - urodziny
print(f"\nIlość dni od urodzin: {roznica.days}")
print(f"To około {roznica.days // 365} lat")

# Dodawanie/odejmowanie czasu
za_tydzien = dzis + timedelta(days=7)
print(f"\nZa tydzień: {za_tydzien}")

za_30_dni = dzis + timedelta(days=30)
print(f"Za 30 dni: {za_30_dni}")

# Formatowanie daty
print(f"\nFormatowanie:")
print(f"ISO: {teraz.isoformat()}")
print(f"Custom: {teraz.strftime('%d/%m/%Y %H:%M:%S')}")
print(f"Polski: {teraz.strftime('%d.%m.%Y')}")
```

## 5. Moduł os - operacje systemowe

```python
import os

# Aktualny katalog roboczy
print(f"Aktualny katalog: {os.getcwd()}")

# Separator ścieżki (różny dla Windows i Linux/Mac)
print(f"Separator: '{os.sep}'")

# Informacje o systemie
print(f"\nSystem: {os.name}")

# Zmienne środowiskowe
print(f"\nZmienne środowiskowe (pierwsze 3):")
for i, (key, value) in enumerate(os.environ.items()):
    if i < 3:
        print(f"  {key}: {value[:50]}..." if len(value) > 50 else f"  {key}: {value}")

# Sprawdzanie istnienia pliku/katalogu
print(f"\nCzy istnieje '/tmp': {os.path.exists('/tmp')}")
print(f"Czy to katalog: {os.path.isdir('/tmp')}")
print(f"Czy to plik: {os.path.isfile('/tmp')}")
```

## 7. Moduł json - praca z JSON

```python
import json

# Python dict -> JSON string
dane_python = {
    "imie": "Jan",
    "wiek": 30,
    "miasto": "Warszawa",
    "hobby": ["programowanie", "czytanie", "sport"],
    "adres": {
        "kod": "80-180",
        "miasto": "Gdańsk"
    }
}

json_string = json.dumps(dane_python, indent=2)
print("JSON string:")
print(json_string)

# JSON string -> Python dict
json_text = '{"nazwa": "Python", "wersja": 3.11, "popularny": true}'
dane = json.loads(json_text)
print(f"\nZ JSON: {dane}")
print(f"Nazwa: {dane['nazwa']}")
print(f"Wersja: {dane['wersja']}")
```

## 8. Moduł collections - zaawansowane struktury danych

```python
from collections import Counter, defaultdict, namedtuple

# Counter - licznik elementów
slowa = "python jest super python jest fajny python".split()
licznik = Counter(slowa)
print("Counter:")
print(licznik)
print(f"Najczęstsze 2: {licznik.most_common(2)}")

# defaultdict - słownik z domyślną wartością
grupy = defaultdict(list)
grupy['owoce'].append('jabłko')
grupy['owoce'].append('banan')
grupy['warzywa'].append('marchew')
print(f"\ndefaultdict: {dict(grupy)}")

# namedtuple - krotka z nazwanymi polami
Punkt = namedtuple('Punkt', ['x', 'y'])
p1 = Punkt(10, 20)
print(f"\nPunkt: {p1}")
print(f"x: {p1.x}, y: {p1.y}")

Osoba = namedtuple('Osoba', ['imie', 'wiek', 'miasto'])
osoba = Osoba('Jan', 30, 'Warszawa')
print(f"\nOsoba: {osoba}")
print(f"Imię: {osoba.imie}")
```

## 9. Moduł itertools - narzędzia do iteracji

```python
from itertools import count, cycle, repeat, combinations, permutations

# count - nieskończony licznik
print("count(10) - pierwsze 5:")
licznik = count(10)
for i in range(5):
    print(next(licznik), end=" ")
print()

# cycle - cykliczne powtarzanie
print("\ncycle(['A', 'B', 'C']) - pierwsze 8:")
cykl = cycle(['A', 'B', 'C'])
for i in range(8):
    print(next(cykl), end=" ")
print()

# repeat - powtarzanie wartości
print("\nrepeat('X', 5):")
print(list(repeat('X', 5)))

# combinations - kombinacje
print("\nKombinacje 2 z [1,2,3,4]:")
for combo in combinations([1, 2, 3, 4], 2):
    print(combo)

# permutations - permutacje
print("\nPermutacje [1,2,3]:")
for perm in permutations([1, 2, 3]):
    print(perm)
```

## 10. Tworzenie własnych modułów

Możesz tworzyć własne moduły, zapisując kod w pliku `.py`.

```python
# Stwórzmy prosty moduł
modul_kod = '''
"""Moduł z funkcjami matematycznymi."""

def kwadrat(x):
    """Zwraca kwadrat liczby."""
    return x ** 2

def szescian(x):
    """Zwraca sześcian liczby."""
    return x ** 3

def silnia(n):
    """Zwraca silnię liczby."""
    if n == 0 or n == 1:
        return 1
    return n * silnia(n - 1)

# Zmienna modułowa
PI = 3.14159

# Kod wykonywany tylko przy bezpośrednim uruchomieniu
if __name__ == "__main__":
    print("Moduł uruchomiony bezpośrednio!")
    print(f"5^2 = {kwadrat(5)}")
'''

# Zapisz moduł
with open('moj_modul.py', 'w') as f:
    f.write(modul_kod)

print("Moduł utworzony: moj_modul.py")
```

```python
# Dodaj ścieżkę do sys.path, aby móc zaimportować
import sys
sys.path.append('')

# Teraz możemy zaimportować nasz moduł
import moj_modul

print(f"kwadrat(7) = {moj_modul.kwadrat(7)}")
print(f"szescian(3) = {moj_modul.szescian(3)}")
print(f"silnia(5) = {moj_modul.silnia(5)}")
print(f"PI = {moj_modul.PI}")
```
